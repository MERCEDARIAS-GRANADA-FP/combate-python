```python
while True:
    operacion = input("¿Qué operación quieres realizar? (sumar/restar/salir): ")

    if operacion == "sumar":
        numero1 = int(input("Introduce un número: "))
        numero2 = int(input("Introduce otro número: "))
        print("Resultado:", numero1 + numero2)

    elif operacion == "restar":
        numero1 = int(input("Introduce un número: "))
        numero2 = int(input("Introduce otro número: "))
        print("Resultado:", numero1 - numero2)

    elif operacion == "salir":
        print("Saliendo del programa.")
        break

    else:
        print("Operación no reconocida. Intenta de nuevo.")
```

Ejercicio 4 Funciones

```python
def factorial(n):

    if n < 0:
        raise ValueError("El número debe ser no negativo.")
    elif n == 0:
        return 1
    else:
        return n * factorial(n - 1)


print(factorial(5))
```

Ejercicio 10 Funciones

```python
def celsius_a_fahrenheit(c):
    return c * 9/5 + 32

def filtrar_por_umbral(lista, umbral):
    return [t for t in lista if t > umbral]

def media(lista):
    return sum(lista) / len(lista) if lista else 0

temperaturas = [12, 25, 30, 18, 22, 35]

temperaturas_f = [celsius_a_fahrenheit(t) for t in temperaturas]
temperaturas_altas = filtrar_por_umbral(temperaturas, 20)
media_temperaturas = media(temperaturas)

print("Temperaturas en Fahrenheit:", temperaturas_f)
print("Temperaturas por encima de 20°C:", temperaturas_altas)
print("Media de temperaturas:", media_temperaturas)
```

Ejercicio 2

```python
class NumeroNegativoError(Exception):
    pass

def solicitar_numero():
    try:
        numero = int(input("Introduce un número entero positivo: "))
        if numero < 0:
            raise NumeroNegativoError("El número no puede ser negativo.")
        return numero
    except ValueError:
        print("Error: Debes introducir un número entero.")
    except NumeroNegativoError as e:
        print(e)

n = solicitar_numero()
if n is not None:
    print(f"Número válido: {n}")
```

Ejericio 4

```python
import math

class ValorNegativoError(Exception):
    pass

def raiz_cuadrada(x):
    try:
        x = float(x)
        if x < 0:
            raise ValorNegativoError("No se puede calcular la raíz cuadrada de un número negativo.")
        return math.sqrt(x)
    except ValueError:
        print("Error: El valor introducido no es un número.")
    except ValorNegativoError as e:
        print(e)

n = input("Introduce un número: ")
resultado = raiz_cuadrada(n)
if resultado is not None:
    print(f"La raíz cuadrada es: {resultado}")
```

Ejercico 8

```python
usuarios = {"juan": "1234", "ana": "abcd"}

def iniciar_sesion(usuario, contraseña):
    if usuario not in usuarios:
        raise KeyError("El usuario no existe.")
    if usuarios[usuario] != contraseña:
        raise ValueError("Contraseña incorrecta.")
    return True

try:
    u = input("Usuario: ")
    c = input("Contraseña: ")
    if iniciar_sesion(u, c):
        print("Inicio de sesión exitoso.")
except KeyError as e:
    print(e)
except ValueError as e:
    print(e)
```
